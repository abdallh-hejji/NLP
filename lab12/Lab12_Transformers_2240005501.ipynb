# Transformers


#### A Transformer is a deep learning architecture designed to process sequences of text using self-attention.

- Transformers are the foundation of many modern NLP models, including BERT, GPT, and T5.

- Before Transformers, we used models such as RNNs and LSTMs for NLP tasks.

## 1. Load a Transformer

We use **DistilBERT**, a  Transformer fine-tuned for positive/negative sentiment classification.

```python
%pip install --upgrade jupyter ipywidgets tqdm
%pip -q install transformers torch matplotlib
```

```python
from transformers import AutoTokenizer, AutoModelForSequenceClassification
import torch

model_name = "distilbert-base-uncased-finetuned-sst-2-english"

tokenizer = AutoTokenizer.from_pretrained(model_name)

model = AutoModelForSequenceClassification.from_pretrained(model_name)
model.eval()

# Look inside the Transformer architecture.
print("Transformer layers:", model.config.n_layers)
print("Attention heads per layer:", model.config.n_heads)
print("Hidden size:", model.config.dim)
```

## 2. Tokenizing the text

```python
sentence = "The movie was not good."

# Tokenization: converts text into token IDs
inputs = tokenizer(sentence, return_tensors="pt")

# [CLS] = start token, [SEP] = end token
tokens = tokenizer.convert_ids_to_tokens(inputs["input_ids"][0])

print("Tokens:", tokens)
```

## 3. Embeddings and self-attention

Each token ID becomes a **vector (embedding)**. Inside each Transformer layer, **self-attention** lets a token use information from other tokens in the same sentence.

```python
import torch

# Enable attention weights
model.set_attn_implementation("eager")

with torch.no_grad():
    embeddings = model.distilbert.embeddings(inputs["input_ids"])
    outputs = model(**inputs, output_attentions=True)

# Attention from the first layer and first head
attention = outputs.attentions[0][0, 0]

print("Embeddings shape:", embeddings.shape)
print("Attention shape:", attention.shape)
print("Tokens:", tokens)
print("Attention weights:\n", attention)
```

## 4. Classification: From Transformer to prediction

The Transformer produces contextual representations. The **classification head** turns them into two scores (*logits*). **Softmax** converts these into probabilities that sum to 1.

```python
# Get prediction scores from the Transformer
with torch.no_grad():
    logits = model(**inputs).logits

# Convert scores into probabilities
probs = torch.softmax(logits, dim=1)[0]

# Display the prediction
prediction = probs.argmax().item()

print("Prediction:", model.config.id2label[prediction])
print("Confidence:", round(probs[prediction].item(), 3))
```

### Task Solution

The following cells load the pretrained `bert-base-uncased` model, inspect its architecture, and extract the attention matrix from the **third layer** and **second attention head**.


```python
# Load the pretrained BERT model and tokenizer
from transformers import AutoTokenizer, AutoModel
import torch

bert_model_name = "bert-base-uncased"
bert_tokenizer = AutoTokenizer.from_pretrained(bert_model_name)
bert_model = AutoModel.from_pretrained(bert_model_name)
bert_model.eval()

# Use the eager attention implementation so attention weights are returned.
# This is also the approach used in Section 3 of this lab.
if hasattr(bert_model, "set_attn_implementation"):
    bert_model.set_attn_implementation("eager")

# 1. Number of Transformer layers and attention heads per layer
num_layers = bert_model.config.num_hidden_layers
num_heads_per_layer = bert_model.config.num_attention_heads

# 2. Total number of attention heads across all layers
total_attention_heads = num_layers * num_heads_per_layer

print("Transformer layers:", num_layers)
print("Attention heads per layer:", num_heads_per_layer)
print("Total attention heads:", total_attention_heads)
```

```python
# 3. Extract attention weights from the third layer and second attention head
sentence = "Transformers use self-attention to understand language."
bert_inputs = bert_tokenizer(sentence, return_tensors="pt")

# Keep inputs and model on the same device (GPU when available, otherwise CPU).
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
bert_model = bert_model.to(device)
bert_inputs = {key: value.to(device) for key, value in bert_inputs.items()}

with torch.no_grad():
    bert_outputs = bert_model(**bert_inputs, output_attentions=True)

# Python uses zero-based indexing:
# third layer -> index 2; second head -> index 1.
third_layer_index = 2
second_head_index = 1
attention_matrix = bert_outputs.attentions[third_layer_index][
    0, second_head_index
]

tokens = bert_tokenizer.convert_ids_to_tokens(bert_inputs["input_ids"][0].detach().cpu().tolist())

print("Tokens:", tokens)
print("Attention matrix shape:", tuple(attention_matrix.shape))
print("Attention weights (rows = query tokens, columns = key tokens):")
print(attention_matrix.detach().cpu())

# Each row is a probability distribution over the tokens it attends to.
print("Row sums (approximately 1.0):")
print(attention_matrix.sum(dim=-1).detach().cpu())
```

```python
# Optional visualization of the selected attention head
import matplotlib.pyplot as plt

attention_to_plot = attention_matrix.detach().cpu().numpy()
plt.figure(figsize=(10, 8))
plt.imshow(attention_to_plot, cmap="viridis", aspect="auto")
plt.xticks(range(len(tokens)), tokens, rotation=45, ha="right")
plt.yticks(range(len(tokens)), tokens)
plt.xlabel("Key tokens (tokens being attended to)")
plt.ylabel("Query tokens (tokens doing the attending)")
plt.title("BERT – Third Transformer Layer, Second Attention Head")
plt.colorbar(label="Attention weight")
plt.tight_layout()
plt.show()
```

### 4. What do the rows and columns represent?

The attention matrix is a square matrix with one row and one column for every token produced by the tokenizer (including special tokens such as `[CLS]` and `[SEP]`).

- **Each row represents a query token**: the token whose representation is being updated.
- **Each column represents a key token**: a token that the query token can look at to gather information.
- **Each cell contains an attention weight**: it indicates how much the query token in that row attends to the key token in that column for this particular attention head.
- **Each row normally sums to approximately 1** because softmax normalizes the attention scores across the key-token positions.

For example, the value in row `use` and column `self-attention` indicates how strongly the token `use` attends to `self-attention` in the selected layer and head. A larger weight means a larger attention contribution from that key position; it does not, by itself, prove a human-style explanation of the model's reasoning.

### Expected architecture results

For `bert-base-uncased`, the model has **12 Transformer layers** and **12 attention heads per layer**, giving **12 × 12 = 144 attention heads** in total. The extracted attention matrix has dimensions `sequence_length × sequence_length`; its exact dimensions depend on how the sentence is tokenized.
